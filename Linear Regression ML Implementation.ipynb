## Lets load the Boston House Pricing Dataset

```python
import pandas as pd
import numpy as np
from types import SimpleNamespace

# Load the Boston Housing dataset
dataset = pd.read_csv("BostonHousing.csv")

# Separate the features and target
feature_names = dataset.columns[:-1].tolist()
data = dataset[feature_names].to_numpy()
target = dataset["medv"].to_numpy()

# Dataset description
description = """
Boston Housing Dataset
----------------------

Number of instances: 506
Number of input features: 13
Target variable: MEDV

The target MEDV represents the median value of owner-occupied
homes, expressed in thousands of US dollars, in the original dataset.

Features:
CRIM    - Per-capita crime rate by town
ZN      - Proportion of residential land zoned for large lots
INDUS   - Proportion of non-retail business acres per town
CHAS    - Charles River dummy variable (1 if tract bounds river, 0 otherwise)
NOX     - Nitric oxide concentration
RM      - Average number of rooms per dwelling
AGE     - Proportion of owner-occupied units built before 1940
DIS     - Weighted distances to employment centers
RAD     - Index of accessibility to radial highways
TAX     - Property tax rate
PTRATIO - Pupil-teacher ratio by town
B       - Historically problematic variable derived from racial demographics
LSTAT   - Percentage of the population classified as lower socioeconomic status

Ethical note:
This dataset has documented ethical limitations. Use it for educational
purposes and interpret its variables and results carefully.
"""

# Create a boston-like object
boston = SimpleNamespace(
    data=data,
    target=target,
    feature_names=np.array(feature_names),
    DESCR=description,
    filename="BostonHousing.csv"
)

# Add a keys() method similar to the original dataset object
boston.keys = lambda: [
    "data", "target", "feature_names", "DESCR", "filename"
]
```

## R square and adjusted R square


Formula

**R^2 = 1 - SSR/SST**


R^2	=	coefficient of determination
SSR	=	sum of squares of residuals
SST	=	total sum of squares


```python
from sklearn.metrics import r2_score
score=r2_score(Y_test,reg_pred)
print(score)
```

Adjusted R2 = 1 – [(1-R2)*(n-1)/(n-k-1)]

where:

R2: The R2 of the model
n: The number of observations
k: The number of predictor variables

```python
#display adjusted R-squared
1 - (1-score)*(len(Y_test)-1)/(len(Y_test)-X_test.shape[1]-1)
```

## RandomForestRegressor

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

model = RandomForestRegressor(
    n_estimators=300,
    min_samples_leaf=2,
    random_state=42
)

model.fit(X_train, Y_train)
predictions = model.predict(X_test)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

print("Random Forest")
print("MAE:", mean_absolute_error(Y_test, predictions))
print("MSE:", mean_squared_error(Y_test, predictions))
print("RMSE:", np.sqrt(mean_squared_error(Y_test, predictions)))
print("R² Score:", r2_score(Y_test, predictions))
```

```python
import pandas as pd

comparison = pd.DataFrame({
    "Actual Price": Y_test.to_numpy(),
    "Predicted Price": predictions
})

print(comparison.head(10))
```

```python
comparison["Residual"] = (
    comparison["Actual Price"] - comparison["Predicted Price"]
)

comparison["Absolute Error"] = comparison["Residual"].abs()

print(comparison.head(10))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(
    Y_test,
    predictions,
    alpha=0.7,
    label="Random Forest predictions"
)

# Perfect prediction reference line
minimum = min(Y_test.min(), predictions.min())
maximum = max(Y_test.max(), predictions.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--",
    label="Perfect predictions"
)

plt.xlabel("Actual Price (thousands of dollars)")
plt.ylabel("Predicted Price (thousands of dollars)")
plt.title("Random Forest: Actual vs Predicted House Prices")
plt.legend()
plt.grid(True)
plt.show()
```
